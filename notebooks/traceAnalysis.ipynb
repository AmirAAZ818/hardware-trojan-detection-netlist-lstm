# Analysis on PCP Traces File

```python
import os
from pathlib import Path
from collections import Counter
from tqdm import tqdm
```

```python
def analyze_traces_file(filepath: str, sample_size: int = 1000):
    """
    Analyze traces file without loading into RAM
    
    Args:
        filepath: Path to traces file
        sample_size: Number of traces to analyze in detail
    """
    print(f"\n{'='*80}")
    print(f"ANALYZING: {Path(filepath).name}")
    print(f"{'='*80}\n")
    
    # File size
    file_size = os.path.getsize(filepath)
    print(f"📁 File size: {file_size / (1024**3):.2f} GB")
    
    # Count lines
    print("\n📊 Counting traces...")
    with open(filepath, 'r', encoding='utf-8') as f:
        num_traces = sum(1 for _ in f)
    print(f"  Total traces: {num_traces:,}")
    
    # Analyze sample
    print(f"\n🔍 Analyzing sample ({sample_size:,} traces)...")
    
    trace_lengths = []
    vocab = Counter()
    
    with open(filepath, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if i >= sample_size:
                break
            
            tokens = line.strip().split()
            trace_lengths.append(len(tokens))
            vocab.update(tokens)
    
    # Statistics
    print(f"\nTrace Statistics (from sample):")
    print(f"  Min length: {min(trace_lengths)}")
    print(f"  Max length: {max(trace_lengths)}")
    print(f"  Avg length: {sum(trace_lengths) / len(trace_lengths):.2f}")
    print(f"  Unique PCP words: {len(vocab):,}")
    
    print(f"\nTop 20 Most Common PCP Words:")
    for word, count in vocab.most_common(20):
        print(f"  {word:40s} {count:5d}")
```

## 100 `.v` Files Analysis

```python
analyze_traces_file(r"E:\University of Kerman\Term 9\VHDL\Paper Implementation\DL_trojan_detection\vocabulary_traces.txt", sample_size=10000)
```
